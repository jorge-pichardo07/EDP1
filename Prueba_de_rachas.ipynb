Aqui hice una función, la llamé rachas, uso el len para tener el numero de muestras, hice la lista IPN vacia, para agregar los 0 y 1, con if verifico si el numero siguiente a la muestra es mayor o menor al anterior y dependiendo de eso agreo un 1 o un 0, despues hago un contador, cada que el número en la lista de 0 y 1 cambie a su contrario (es decir, despues de los 0, aparezaca un 1) sumo 1 al conteo


```python
def rachas(X):
  n= len(X)
  IPN = []
  conteos = 1
  for i in range(0,len(X)-1):
    if X[i+1] < X[i]:
      IPN.append(0)
    else:
      IPN.append(1)
  for i in range(len(IPN)-1):
    if IPN[i] != IPN[i+1] :
      conteos += 1
    else:
      conteos += 0.
  print(f'El número de rachas contados fue C_0={conteos}')
  z_n = stats.norm.ppf(1-0.05/2)
  print(f'Nuestro estadístico para el 95% de confianza es Z_n = {z_n:.2f} y -{z_n:.2f}')
  M= (2*n-1)/3
  Sig= (16*n-29)/90
  z_0=(conteos-M)/Sig
  print('Usamos el estadístico:')
  display(Math(r"z_0 = \frac{C_0 - M_{C0}}{\sigma^2}"))
  if abs(z_0) < z_n:
    print(f'Notemos que -{z_n:.2f} < {z_0:.2f} < {z_n:.2f}')
    print('No se rechaza la prueba de hipótesis, los datos de la muestra son independientes')
  else:
    print(f'Notemos que {z_n:.2f} cae en la zona de rechazo')
    print('Se rechaza la prueba de hipótesis, los datos de la muestra NO son independientes')
```

Al final al usar la función, esta devuelve varios print, para especificar el numero de conteos, el estadistico estandarizado que usamos, y el estadistico de prueba que usaremos, en este caso es:
$$z_0 = \frac{C_0 - M_{C0}}{\sigma^2}$$

al final, muestro que en efecto el estadistico cae en la zona de no rechazo de la prueba de hipótesis, o en caso contrario, el estadistico cae en la zona de rechazo, mostrando que la la muestra es o no es independiente.
